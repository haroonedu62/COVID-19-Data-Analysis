**Import Libraries**

```python
# Import NumPy for numerical calculations
import numpy as np

# Import Pandas for data manipulation and analysis
import pandas as pd

# Import Matplotlib for data visualization
import matplotlib.pyplot as plt

# Import Seaborn for attractive statistical visualizations
import seaborn as sns

# Set the visualization style
sns.set_style("whitegrid")

# Display plots inside Google Colab
%matplotlib inline

print("Libraries imported successfully!")
```

**Upload Dataset**

```python
df = pd.read_csv("/content/COVID DEATHS.csv")

df.head()
```

**Understand the Dataset**

```python
# Display the number of rows and columns
print("Dataset Shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display information about the dataset
print("\nDataset Information:")
df.info()

# Display statistical summary of numerical columns
print("\nStatistical Summary:")
df.describe()

# Display the first five rows
display(df.head())

# Display the last five rows
display(df.tail())
```

**Data Cleaning**

```python
# Create a copy of the original dataset
covid = df.copy()

# Convert the date column into datetime format
covid["date"] = pd.to_datetime(covid["date"], errors="coerce")

# Sort the dataset by country and date
covid = covid.sort_values(
    by=["location", "date"]
).reset_index(drop=True)

# Display the updated dataset
covid.head()
```

**Remove duplicate records**

```python
# Check the number of duplicate rows
print("Duplicate Rows Before Cleaning:", covid.duplicated().sum())

# Remove duplicate rows
covid = covid.drop_duplicates()

# Check duplicates after cleaning
print("Duplicate Rows After Cleaning:", covid.duplicated().sum())
```

**Check missing values**

```python
# Count missing values in each column
missing_values = covid.isnull().sum()

# Calculate missing-value percentages
missing_percentage = (
    covid.isnull().sum() / len(covid)
) * 100

# Create a missing-value summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

# Sort the summary by missing values
missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

# Display the summary
missing_summary.head(20)
```

**Explore Countries and Dates**

```python
# Display the number of unique locations
print("Total Unique Locations:", covid["location"].nunique())

# Display the first 20 locations
print("\nSample Locations:")
print(covid["location"].unique()[:20])

# Find the earliest date
print("\nEarliest Date:", covid["date"].min())

# Find the latest date
print("Latest Date:", covid["date"].max())
```

**Count records by continent**

```python
# Check whether the continent column exists
if "continent" in covid.columns:

    # Count records for each continent
    continent_counts = covid["continent"].value_counts(dropna=False)

    # Display the results
    print(continent_counts)
else:
    print("Continent column is not available.")
```

# **Exploratory Data Analysis**

**Global COVID-19 Cases Over Time**

```python
# Select rows where the location is World
world_data = covid[covid["location"] == "World"].copy()

# Check whether World-level data is available
if world_data.empty:
    print("World-level data is not available in this dataset.")
else:
    # Sort the data by date
    world_data = world_data.sort_values("date")

    # Create a line chart
    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=world_data,
        x="date",
        y="total_cases"
    )

    plt.title("Global COVID-19 Total Cases Over Time")
    plt.xlabel("Date")
    plt.ylabel("Total Cases")
    plt.xticks(rotation=45)
    plt.show()
```

**Global COVID-19 Deaths Over Time**

```python
# Check whether World-level data is available
if world_data.empty:
    print("World-level data is not available.")
else:
    # Create a line chart for total deaths
    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=world_data,
        x="date",
        y="total_deaths"
    )

    plt.title("Global COVID-19 Total Deaths Over Time")
    plt.xlabel("Date")
    plt.ylabel("Total Deaths")
    plt.xticks(rotation=45)
    plt.show()
```

**Daily New Cases and Deaths**

```python
# Check whether World-level data is available
if world_data.empty:
    print("World-level data is not available.")
else:
    # Create a figure for daily new cases
    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=world_data,
        x="date",
        y="new_cases"
    )

    plt.title("Global Daily New COVID-19 Cases")
    plt.xlabel("Date")
    plt.ylabel("New Cases")
    plt.xticks(rotation=45)
    plt.show()

    # Create a separate figure for daily new deaths
    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=world_data,
        x="date",
        y="new_deaths"
    )

    plt.title("Global Daily New COVID-19 Deaths")
    plt.xlabel("Date")
    plt.ylabel("New Deaths")
    plt.xticks(rotation=45)
    plt.show()
```

**Top 10 Countries by Total Cases**

```python
# Select country-level records
# Exclude aggregate locations such as World and continents
country_data = covid[
    covid["continent"].notna()
].copy()

# Get the latest available record for each country
latest_country_data = (
    country_data
    .sort_values("date")
    .groupby("location")
    .tail(1)
)

# Sort countries by total cases
top_cases = latest_country_data[
    ["location", "total_cases"]
].sort_values(
    by="total_cases",
    ascending=False
).head(10)

# Display the top 10 countries
top_cases
```

**Visualize top countries**

```python
# Create a horizontal bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_cases,
    x="total_cases",
    y="location"
)

plt.title("Top 10 Countries by Total COVID-19 Cases")
plt.xlabel("Total Cases")
plt.ylabel("Country")
plt.show()
```

**Top 10 Countries by Total Deaths**

```python
# Sort countries by total deaths
top_deaths = latest_country_data[
    ["location", "total_deaths"]
].sort_values(
    by="total_deaths",
    ascending=False
).head(10)

# Display the top 10 countries
top_deaths
```

**Visualize top countries by deaths**

```python
# Create a horizontal bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_deaths,
    x="total_deaths",
    y="location"
)

plt.title("Top 10 Countries by Total COVID-19 Deaths")
plt.xlabel("Total Deaths")
plt.ylabel("Country")
plt.show()
```

**COVID-19 Cases Per Million People**

```python
# Check whether the required column exists
if "total_cases_per_million" in covid.columns:

    # Get the latest record for each country
    latest_per_million = (
        country_data
        .sort_values("date")
        .groupby("location")
        .tail(1)
    )

    # Find the top 10 countries by cases per million
    top_cases_per_million = latest_per_million[
        ["location", "total_cases_per_million"]
    ].sort_values(
        by="total_cases_per_million",
        ascending=False
    ).head(10)

    # Display the results
    display(top_cases_per_million)

    # Create a chart
    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=top_cases_per_million,
        x="total_cases_per_million",
        y="location"
    )

    plt.title("Top 10 Countries by COVID-19 Cases Per Million")
    plt.xlabel("Cases Per Million")
    plt.ylabel("Country")
    plt.show()

else:
    print("total_cases_per_million column is not available.")
```

**NumPy Statistical Analysis**

```python
# Select valid total case values
case_values = covid["total_cases"].dropna().to_numpy()

# Select valid total death values
death_values = covid["total_deaths"].dropna().to_numpy()

# Calculate statistical values using NumPy
average_cases = np.mean(case_values)
median_cases = np.median(case_values)
maximum_cases = np.max(case_values)
minimum_cases = np.min(case_values)
standard_deviation_cases = np.std(case_values)

# Display the results
print("NUMPY COVID-19 ANALYSIS")
print("Average Total Cases:", round(average_cases, 2))
print("Median Total Cases:", round(median_cases, 2))
print("Maximum Total Cases:", round(maximum_cases, 2))
print("Minimum Total Cases:", round(minimum_cases, 2))
print("Standard Deviation:", round(standard_deviation_cases, 2))
```

**NumPy analysis of daily cases**

```python
# Select valid daily new case values
daily_cases = covid["new_cases"].dropna().to_numpy()

# Calculate daily case statistics
average_daily_cases = np.mean(daily_cases)
median_daily_cases = np.median(daily_cases)
maximum_daily_cases = np.max(daily_cases)

# Display the results
print("Average Daily New Cases:", round(average_daily_cases, 2))
print("Median Daily New Cases:", round(median_daily_cases, 2))
print("Maximum Daily New Cases:", round(maximum_daily_cases, 2))
```

**Correlation Analysis**

```python
# Select numerical columns available in the dataset
correlation_columns = [
    "total_cases",
    "new_cases",
    "total_deaths",
    "new_deaths",
    "population"
]

# Keep only columns that exist in the dataset
available_columns = [
    column for column in correlation_columns
    if column in covid.columns
]

# Calculate the correlation matrix
correlation_matrix = covid[available_columns].corr()

# Display the correlation matrix
correlation_matrix
```

**Correlation Heatmap**

```python
# Create a heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("COVID-19 Numerical Features Correlation Heatmap")
plt.show()
```

**Merge Deaths and Vaccination Data**

```python
# Load the deaths dataset
deaths_df = pd.read_csv("COVID DEATHS.csv")

# Load the vaccinations dataset
vaccinations_df = pd.read_csv("/content/COVID DEATHS.csv")

# Display the column names of both datasets
print("Deaths Columns:")
print(deaths_df.columns.tolist())

print("\nVaccination Columns:")
print(vaccinations_df.columns.tolist())
```

**Clean date columns**

```python
# Convert date columns into datetime format
deaths_df["date"] = pd.to_datetime(
    deaths_df["date"],
    errors="coerce"
)

vaccinations_df["date"] = pd.to_datetime(
    vaccinations_df["date"],
    errors="coerce"
)
```

**Merge the datasets**

```python
# Define the common columns used for merging
merge_columns = ["iso_code", "location", "date"]

# Merge deaths and vaccination datasets
covid_merged = pd.merge(
    deaths_df,
    vaccinations_df,
    on=merge_columns,
    how="left",
    suffixes=("_deaths", "_vaccinations")
)

# Display the merged dataset
covid_merged.head()
```

**Vaccination Progress Analysis**

```python
# Check whether the vaccination column exists
if "people_vaccinated" in covid.columns:

    # Select World-level vaccination data
    world_vaccination = covid[
        covid["location"] == "World"
    ].copy()

    # Sort data by date
    world_vaccination = world_vaccination.sort_values("date")

    # Create a line chart
    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=world_vaccination,
        x="date",
        y="people_vaccinated"
    )

    plt.title("Global COVID-19 Vaccination Progress")
    plt.xlabel("Date")
    plt.ylabel("People Vaccinated")
    plt.xticks(rotation=45)
    plt.show()

else:
    print("Vaccination column is not available in the main dataset.")
```

**Top Countries by Vaccination Percentage**

**COVID-19 Cases vs. Deaths**

```python
# Select rows with valid case and death values
scatter_data = covid[
    ["total_cases", "total_deaths"]
].dropna()

# Create a scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=scatter_data,
    x="total_cases",
    y="total_deaths",
    alpha=0.5
)

plt.title("Relationship Between Total COVID-19 Cases and Deaths")
plt.xlabel("Total Cases")
plt.ylabel("Total Deaths")
plt.show()
```
